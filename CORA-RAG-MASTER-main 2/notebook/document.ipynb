```python
### textloader -> para leer el texto 

from langchain_community.document_loaders import TextLoader

loader=TextLoader("../data/text_files/history_ia.txt", encoding="utf-8") ##crear el textloader para que lea el contenido del document
document=loader.load() #Lo lee 
print(document)

```

```python
## Directory Loader

from langchain_community.document_loaders import TextLoader, DirectoryLoader

## load all txt files from directory ## 

dir_loader = DirectoryLoader(
    "../data/text_files",
    glob= "**/*.txt", ## Carga UNICAMENTE los archivos de .txt
    loader_cls= TextLoader, ## Con textloader, lee cada archivo individualmente 
    loader_kwargs= {'encoding' : 'utf-8'}, ## el encoding 
    show_progress=False ## No muestra la barra de carga
)

documents = dir_loader.load() ##carga todos los archivos y los convierte en lista para documen
documents
```

```python
## PARA LOS PDFS !!! ##

from langchain_community.document_loaders import PyMuPDFLoader

## load all txt files from directory ## 

dir_loader = DirectoryLoader(
    "../data/pdf",
    glob= "**/*.pdf", ## Carga UNICAMENTE los archivos de .txt
    loader_cls= PyMuPDFLoader, ## Con textloader, lee cada archivo individualmente 
    ##loader_kwargs= {'encoding' : 'utf-8'}, !!!! Aca no es necesario usar el encoding ya que por la naturaleza de la libreria, viene por defalt 
    show_progress=False ## No muestra la barra de carga
)

pdf_documents=dir_loader.load()
pdf_documents

```
